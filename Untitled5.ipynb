```python
import pandas as pd

# -----------------------------
# CONFIG
# -----------------------------
FILE_PATH = "D:/Tushar/Capacity_Load_Calculation_Master.xlsx"
TARGET_HOURS = 22   # FULL UTILIZATION TARGET

# -----------------------------
# LOAD DATA
# -----------------------------
df = pd.read_excel(FILE_PATH)

# -----------------------------
# FILTER ONLY 120T MACHINES
# -----------------------------
df_120 = df[df["Tonnage"] == 120].copy()

# -----------------------------
# CALCULATE LOAD (HOURS / DAY)
# -----------------------------
df_120["Load_Hours"] = (
    df_120["Daily Demand"] * df_120["Cycle Time"]
) / 3600

# =========================================================
# 1️⃣ MACHINE-WISE RAW PLAN
# =========================================================
print("\n" + "=" * 90)
print("📋 MACHINE-WISE RAW PRODUCTION PLAN")
print("=" * 90)

for machine_id, machine_df in df_120.groupby("Machine Id"):
    print(f"\n🛠 MACHINE : {machine_id}")
    print(
        machine_df[
            ["Part No", "Color", "Daily Demand", "Cycle Time", "Load_Hours"]
        ].to_string(index=False)
    )

# =========================================================
# 2️⃣ RAW LOAD ON EACH MACHINE
# =========================================================
machine_load = (
    df_120.groupby("Machine Id")["Load_Hours"]
    .sum()
    .reset_index()
)

machine_load["Unbalanced_Load"] = machine_load["Load_Hours"]
machine_load["Balanced_Load"] = machine_load["Load_Hours"]

print("\n" + "=" * 90)
print("🔴 RAW LOAD ON EACH MACHINE")
print("=" * 90)

raw_display = machine_load.copy()
raw_display["Utilization_%"] = (raw_display["Unbalanced_Load"] / TARGET_HOURS) * 100
raw_display["Status"] = raw_display["Unbalanced_Load"].apply(
    lambda x: "OVERLOADED ❌" if x > TARGET_HOURS else "OK ✅"
)

print(
    raw_display[
        ["Machine Id", "Unbalanced_Load", "Utilization_%", "Status"]
    ].to_string(index=False)
)

# =========================================================
# 3️⃣ BALANCING LOGIC (TARGET = 22 HOURS)
# =========================================================
overloaded = machine_load[machine_load["Balanced_Load"] > TARGET_HOURS]
underloaded = machine_load[machine_load["Balanced_Load"] < TARGET_HOURS]

overloaded = overloaded.sort_values(by="Balanced_Load", ascending=False)
underloaded = underloaded.sort_values(by="Balanced_Load")

for o_idx, o_row in overloaded.iterrows():

    excess = o_row["Balanced_Load"] - TARGET_HOURS

    for u_idx, u_row in underloaded.iterrows():

        free = TARGET_HOURS - u_row["Balanced_Load"]

        if excess <= 0:
            break
        if free <= 0:
            continue

        shift = min(excess, free)

        machine_load.loc[o_idx, "Balanced_Load"] -= shift
        machine_load.loc[u_idx, "Balanced_Load"] += shift

        excess -= shift

    if excess > 0:
        print(
            f"⚠️ Remaining overload {round(excess,2)} hrs on machine "
            f"{o_row['Machine Id']} (plant capacity insufficient)"
        )

# =========================================================
# 4️⃣ MACHINE-WISE BALANCED PLAN (UNCHANGED PART VIEW)
# =========================================================
print("\n" + "=" * 90)
print("📋 MACHINE-WISE BALANCED PLAN (PART VIEW – SAME)")
print("=" * 90)

for machine_id, machine_df in df_120.groupby("Machine Id"):
    print(f"\n🛠 MACHINE : {machine_id}")
    print(
        machine_df[
            ["Part No", "Color", "Daily Demand", "Cycle Time", "Load_Hours"]
        ].to_string(index=False)
    )

# =========================================================
# 5️⃣ BALANCED LOAD ON EACH MACHINE
# =========================================================
print("\n" + "=" * 90)
print("🟢 BALANCED LOAD ON EACH MACHINE (TARGET = 22 HRS)")
print("=" * 90)

balanced_display = machine_load.copy()
balanced_display["Utilization_%"] = (
    balanced_display["Balanced_Load"] / TARGET_HOURS
) * 100

balanced_display["Status"] = balanced_display["Balanced_Load"].apply(
    lambda x: "OK ✅" if x <= TARGET_HOURS else "OVERLOADED ❌"
)

print(
    balanced_display[
        ["Machine Id", "Balanced_Load", "Utilization_%", "Status"]
    ].to_string(index=False)
)
```

```python
import pandas as pd

# -----------------------------
# CONFIG
# -----------------------------
FILE_PATH = "D:/Tushar/Capacity_Load_Calculation_Master.xlsx"
SHIFTS = 2
HOURS_PER_SHIFT = 11
AVAILABLE_HOURS_PER_DAY = SHIFTS * HOURS_PER_SHIFT  # 22 hours

# -----------------------------
# LOAD DATA
# -----------------------------
df = pd.read_excel(FILE_PATH)

# -----------------------------
# FILTER ONLY 120 TON MACHINES
# -----------------------------
df = df[df["Tonnage"] == 120].copy()

# -----------------------------
# CALCULATE LOAD (HOURS + MINUTES)
# -----------------------------
df["Load_Hours_Per_Day"] = (df["Daily Demand"] * df["Cycle Time"]) / 3600
df["Load_Minutes_Per_Day"] = df["Load_Hours_Per_Day"] * 60

# -----------------------------
# KEEP UNBALANCED COPY
# -----------------------------
unbalanced_df = df.copy()

# -----------------------------
# INITIAL MACHINE LOAD
# -----------------------------
machine_load = (
    df.groupby("Machine Id")["Load_Hours_Per_Day"]
    .sum()
    .to_dict()
)

# -----------------------------
# BALANCING (PART-LEVEL LOAD SHIFT)
# -----------------------------
balanced_rows = []
shift_log = []   # 👈 THIS IS NEW

overloaded_machines = sorted(machine_load.items(), key=lambda x: x[1], reverse=True)
underloaded_machines = sorted(machine_load.items(), key=lambda x: x[1])

for o_machine, o_load in overloaded_machines:

    if o_load <= AVAILABLE_HOURS_PER_DAY:
        continue

    parts = df[df["Machine Id"] == o_machine]

    for idx, part in parts.iterrows():

        excess = machine_load[o_machine] - AVAILABLE_HOURS_PER_DAY
        if excess <= 0:
            break

        part_load = part["Load_Hours_Per_Day"]

        for u_machine, _ in underloaded_machines:

            if u_machine == o_machine:
                continue

            free = AVAILABLE_HOURS_PER_DAY - machine_load[u_machine]
            if free <= 0:
                continue

            shift = min(part_load, excess, free)

            # Reduce load on original machine
            df.loc[idx, "Load_Hours_Per_Day"] -= shift
            df.loc[idx, "Load_Minutes_Per_Day"] -= shift * 60
            machine_load[o_machine] -= shift

            # Create shifted row
            new_row = part.copy()
            new_row["Machine Id"] = u_machine
            new_row["Load_Hours_Per_Day"] = shift
            new_row["Load_Minutes_Per_Day"] = shift * 60
            balanced_rows.append(new_row)

            # -----------------------------
            # LOG THE SHIFT 🔥
            # -----------------------------
            shift_log.append({
                "From_Machine": o_machine,
                "To_Machine": u_machine,
                "Part_No": part["Part No"],
                "Shifted_Hours": round(shift, 2),
                "Shifted_Minutes": round(shift * 60, 1)
            })

            machine_load[u_machine] += shift
            excess -= shift

            if excess <= 0:
                break

# -----------------------------
# FINAL BALANCED DATAFRAME
# -----------------------------
balanced_df = pd.concat([df, pd.DataFrame(balanced_rows)], ignore_index=True)
balanced_df = balanced_df[balanced_df["Load_Hours_Per_Day"] > 0]

# =========================================================
# 🔴 UNBALANCED MACHINE-WISE PLAN
# =========================================================
print("\n" + "#" * 80)
print("🔴 UNBALANCED MACHINE-WISE PLAN")
print("#" * 80)

for machine_id, machine_df in unbalanced_df.groupby("Machine Id"):

    total_load = machine_df["Load_Hours_Per_Day"].sum()
    utilization = (total_load / AVAILABLE_HOURS_PER_DAY) * 100
    status = "OVERLOADED ❌" if total_load > AVAILABLE_HOURS_PER_DAY else "OK ✅"

    print("\n" + "=" * 60)
    print(f"🛠 MACHINE ID        : {machine_id}")
    print(f"⏱ TOTAL LOAD        : {round(total_load, 2)} hours/day")
    print(f"🕒 AVAILABLE HOURS  : {AVAILABLE_HOURS_PER_DAY} hours/day")
    print(f"📊 UTILIZATION      : {round(utilization, 2)} %")
    print(f"🚦 STATUS           : {status}")
    print("=" * 60)

    print(
        machine_df[
            [
                "Machine Id",
                "Part No",
                "Color",
                "Daily Demand",
                "Cycle Time",
                "Load_Hours_Per_Day",
                "Load_Minutes_Per_Day",
            ]
        ].sort_values(by="Part No").to_string(index=False)
    )

# =========================================================
# 🟢 BALANCED MACHINE-WISE PLAN
# =========================================================
print("\n" + "#" * 80)
print("🟢 BALANCED MACHINE-WISE PLAN")
print("#" * 80)

for machine_id, machine_df in balanced_df.groupby("Machine Id"):

    total_load = machine_df["Load_Hours_Per_Day"].sum()
    utilization = (total_load / AVAILABLE_HOURS_PER_DAY) * 100
    status = "OK ✅" if total_load <= AVAILABLE_HOURS_PER_DAY else "OVERLOADED ❌"

    print("\n" + "=" * 60)
    print(f"🛠 MACHINE ID        : {machine_id}")
    print(f"⏱ TOTAL LOAD        : {round(total_load, 2)} hours/day")
    print(f"🕒 AVAILABLE HOURS  : {AVAILABLE_HOURS_PER_DAY} hours/day")
    print(f"📊 UTILIZATION      : {round(utilization, 2)} %")
    print(f"🚦 STATUS           : {status}")
    print("=" * 60)

    print(
        machine_df[
            [
                "Machine Id",
                "Part No",
                "Color",
                "Daily Demand",
                "Cycle Time",
                "Load_Hours_Per_Day",
                "Load_Minutes_Per_Day",
            ]
        ].sort_values(by="Part No").to_string(index=False)
    )

# =========================================================
# 📦 PART LOAD SHIFT DETAILS (NEW & CLEAR)
# =========================================================
print("\n" + "#" * 80)
print("📦 PART LOAD SHIFT DETAILS (WHO MOVED WHAT & WHERE)")
print("#" * 80)

if shift_log:
    shift_df = pd.DataFrame(shift_log)
    print(shift_df.to_string(index=False))
else:
    print("✅ No load shifting was required. All machines were already balanced.")
```
