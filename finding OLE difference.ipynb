```python
import pandas as pd

# =========================
# LOAD CSV
# =========================

df = pd.read_csv("machine_oee_data.csv")

# =========================
# FIX DATE FORMAT
# =========================

df["Date"] = pd.to_datetime(df["Date"]).dt.date

# =========================
# PIVOT SHIFTS INTO COLUMNS
# =========================

pivot_df = df.pivot_table(
    index=["Date", "Machine"],
    columns="Shift",
    values="OEE",
    aggfunc="first"
).reset_index()

# =========================
# RENAME SHIFT COLUMNS
# =========================

pivot_df.columns = [
    f"OEE Shift {col}" if col not in ["Date", "Machine"] else col
    for col in pivot_df.columns
]

# =========================
# CALCULATE DIFFERENCE
# =========================

shift_cols = [c for c in pivot_df.columns if "OEE Shift" in c]

pivot_df["Difference"] = pivot_df[shift_cols].max(axis=1) - pivot_df[shift_cols].min(axis=1)

# =========================
# SORT BY DATE AND MACHINE
# =========================

pivot_df = pivot_df.sort_values(by=["Date", "Machine"])

# =========================
# SAVE FILE
# =========================

pivot_df.to_csv("machine_oee_shift_comparison.csv", index=False)

print("File created successfully")
```
