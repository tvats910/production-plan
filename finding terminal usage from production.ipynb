```python
import pandas as pd

# =========================
# READ FILES
# =========================

# File 1: part-terminal mapping
df_terminal = pd.read_excel("part_terminal.xlsx")

# File 2: production data
df_prod = pd.read_excel("production.xlsx")

# =========================
# MERGE PRODUCTION DATA
# =========================

df = df_terminal.merge(df_prod, on="part", how="left")

# Replace missing production with 0
df["production"] = df["production"].fillna(0)

# =========================
# TERMINAL COLUMNS
# =========================

terminal_cols = [col for col in df.columns if "terminal" in col.lower()]

# =========================
# CALCULATE TERMINAL USAGE
# =========================

terminal_usage = {}

for _, row in df.iterrows():

    production_qty = row["production"]

    for col in terminal_cols:

        terminal = row[col]

        # Ignore empty cells
        if pd.notna(terminal):

            if terminal not in terminal_usage:
                terminal_usage[terminal] = 0

            terminal_usage[terminal] += production_qty

# =========================
# CREATE OUTPUT
# =========================

output_df = pd.DataFrame(
    terminal_usage.items(),
    columns=["terminal", "qty_used"]
)

# Sort terminals
output_df = output_df.sort_values("terminal")

# Save output
output_df.to_excel("terminal_usage_output.xlsx", index=False)

print(output_df)
```
