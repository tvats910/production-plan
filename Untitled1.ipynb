```python
import pandas as pd
import numpy as np

# =====================================================
# CONFIG
# =====================================================
FILE_PATH = "Copy of Master Data_290102026 2 - Copy.xlsx"
CHANGEOVER_TIME = 40
DAILY_MACHINE_CAPACITY = 1320

# =====================================================
# STEP 1: LOAD DATA
# =====================================================

df = pd.read_excel(FILE_PATH)

# Clean column names
df.columns = df.columns.str.strip()

# =====================================================
# STEP 2: CALCULATE DEMAND PER CHILD PART
# =====================================================

# Effective daily demand = Daily Plan × Sub Count
df["effective_daily_demand"] = (
    df["Daily Plan"] * df["Sub Count"]
)

grouped = df.groupby("Child Part", sort=False)

child_df = grouped.agg({
    "effective_daily_demand": "sum",
    "Inventory_25": "first",
    " Minimum Quantity": "first",
    "Cycle Time": "first",
    "Vertical Machines": lambda x: list(
        set(",".join(x.astype(str)).split(","))
    )
}).reset_index()

child_df.rename(columns={
    "Child Part": "part",
    "effective_daily_demand": "daily_demand",
    "Inventory_25": "inventory",
    " Minimum Quantity": "min_qty",
    "Cycle Time": "cycle_time",
    "Vertical Machines": "eligible_machines"
}, inplace=True)

# =====================================================
# STEP 3: NET REQUIRED QTY (YOUR LOGIC)
# =====================================================

child_df["net_required_qty"] = (
    child_df["daily_demand"]
    + child_df["min_qty"]
    - child_df["inventory"]
)

child_df["net_required_qty"] = child_df["net_required_qty"].clip(lower=0)

# =====================================================
# STEP 4: BUILD MACHINE CAPACITY MAP
# =====================================================

machines = (
    child_df["eligible_machines"]
    .explode()
    .unique()
)

machine_capacity = {
    m.strip(): DAILY_MACHINE_CAPACITY
    for m in machines if m.strip()
}

# =====================================================
# STEP 5: BASELINE OPTIMIZER
# =====================================================

def run_daily_optimizer(parts_df, machine_capacity):
    df = parts_df.copy()

    df["inventory_days"] = (
        df["inventory"] / df["daily_demand"]
    ).replace([np.inf, -np.inf], 99).fillna(99)

    df["production_time"] = (
        df["net_required_qty"] * df["cycle_time"]
    )

    df["setup_efficiency"] = (
        df["production_time"] / CHANGEOVER_TIME
    )

    df["inventory_risk"] = np.maximum(
        0, 2 - df["inventory_days"]
    )

    df["must_run"] = (
        (df["inventory_days"] < 1) |
        (df["inventory"] <= 0)
    )

    max_demand = df["daily_demand"].max()
    df["priority"] = (
        2 * df["inventory_risk"]
        + df["setup_efficiency"]
        + (df["daily_demand"] / max_demand)
    )

    df["eligible_today"] = (
        df["must_run"] |
        (df["setup_efficiency"] >= 1)
    )

    schedule = []
    deferred = []

    machine_capacity = machine_capacity.copy()

    for _, row in df.sort_values("priority", ascending=False).iterrows():

        qty_left = row["net_required_qty"]

        if not row["eligible_today"]:
            deferred.append({
                "part": row["part"],
                "qty_left": qty_left,
                "reason": "LOW_SETUP_EFFICIENCY"
            })
            continue

        for machine in row["eligible_machines"]:

            machine = machine.strip()
            if qty_left <= 0 or machine not in machine_capacity:
                break

            remaining = machine_capacity[machine]

            if remaining <= CHANGEOVER_TIME:
                continue

            max_qty_today = int(
                (remaining - CHANGEOVER_TIME) / row["cycle_time"]
            )

            if max_qty_today <= 0:
                continue

            qty_today = min(qty_left, max_qty_today)

            prod_time = qty_today * row["cycle_time"]
            total_time = prod_time + CHANGEOVER_TIME

            machine_capacity[machine] -= total_time
            qty_left -= qty_today

            schedule.append({
                "machine": machine,
                "part": row["part"],
                "qty": qty_today,
                "prod_time_min": round(prod_time, 1),
                "changeover_min": CHANGEOVER_TIME,
                "reason": "MUST_RUN" if row["must_run"] else "PRIORITY"
            })

        if qty_left > 0:
            deferred.append({
                "part": row["part"],
                "qty_left": qty_left,
                "reason": "CAPACITY_LIMIT"
            })

    return pd.DataFrame(schedule), pd.DataFrame(deferred)

# =====================================================
# STEP 6: RUN
# =====================================================

schedule_df, deferred_df = run_daily_optimizer(
    child_df,
    machine_capacity
)

print("\n===== FINAL DAILY SCHEDULE =====")
print(schedule_df)

print("\n===== DEFERRED PARTS =====")
print(deferred_df)
```
