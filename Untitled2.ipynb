```python
def normalize_columns(df):
    df = df.copy()
    df_columns = (
        df.columns
        .astype(str)
        .str.strip()
        .str.lower()
        .str.replace(" ", "-")
        .str.replace(".", "", regex=False)
    )
    return df


def norm(x):
    if pd.isna(x):
        return ""

    return str(x).strip().lower()
```

```python
df_master = normalize_columns(df_master)
df_inventory = normalize_columns(df_inventory)
df_sheet10 = normalize_columns(df_sheet10)
df_fg = normalize_columns(df_fg)
df_prod = normalize_columns(df_prod)
df_tool = normalize_columns(df_tool)
```

```python
def best_match(df, series, part, series_col=None, part_col=None):
    series_n = norm(series)
    part_n = norm(part)

    best_row = None
    best_score = 0

    for _,r in df.iterrows():
        series_match = False
        part_match = False

        if series_col and series_col in df.columns:
            series_match = norm(r[series_col]) == series_n

        if part_col and part_col in df.columns:
            part_match = norm(r[part_col]) == part_n


        if series_match and part_match:
            score = 3

        elif part_match:
            score = 2

        elif series_match:
            score = 1
        else:
            continue

        if score > best_score:
            best_score = score
            best_row = r


    return best_row
```

```python
def get_fg_stock(master_row):
    series = master_row["Series"]
    part = master_row["Part"]

    r = best_match(df_inventory, series, part, part_col="Material")
    if r is not None and not pd.isna(r["Unrestricted"]):
        return r["Unrestricted"]

    if not pd.isna(master_row["FG Quantity"]):
        return master_row["FG Quantity"]

    r = best_match(df_sheet10, series, part, series_col="Material")
    if r is not None and not pd.isna(r["Unrestricted"]):
        return r["Unrestricted"]

    r = best_match(df_fg, series, part, series_col="Series", part_col="Part")
    if r is not None and  not pd.isna(r["FG Quantity"]):
        return r["FG Quantity"]


    return 0
```

```python
def get_machine(series, part):
    r = best_match(df_tool, series, part, part_col="Part No.")
    if r is not None:
        return r["Machine No."]

    return None
```

```python
def get_time_required(series, part, qty):
    r = best_match(df_prod, series, part, part_col="Part No.")
    if r is None:
        return None

    pph = r.get("Part Per Hour")
    if not pd.isna(pph) and pph > 0:
        return qty / pph

    return None
```

```python
plan = []

for  i, row_tuple in enumerate(df_master.iterrows()):

    _, row = row_tuple
    if i % 5000 == 0:
        print(f"processed {i} rows")

    
    series = row["Series"]
    part = row["Part"]


    monthly_req = row['Monthely Requirement-jan'] if not pd.isna(row['Monthely Requirement-jan']) else 0
    min_req     = row["Minimum Requirement"] if not pd.isna(row["Minimum Requirement"]) else 0
    dispatch    = row["Dispatch"] if not pd.isna(row["Dispatch"]) else 0


    fg_stock = get_fg_stock(row)
    available_fg = fg_stock - dispatch

    net_req = monthly_req + min_req - available_fg
    if net_req <= 0:
        continue

    daily_demand = monthly_req / 28 if monthly_req > 0 else 0

    if available_fg < daily_demand:
        planned_qty = daily_demand

    elif daily_demand <= 50:
        planned_qty = min(5 * daily_demand, net_req)

    else:
        planned_qty = net_req

    if planned_qty <= 0:
        continue

    machine = get_machine(series, part)
    if machine is None:
        continue


    time_required = get_time_required(series, part, planned_qty)
    if time_required is None:
        print("Skipped (no cycle time): ", part)
        continue

    plan.append({
            machine, 
            part, 
            int(round(planned_qty)), 
            int(round(time_required, 2))
        })
```

```python
df_plan = pd.DataFrame(
    plan,
    columns = ["Machine", "Part", "Qty", "Time Required (hrs)"]
)

df_plan = df_plan.sort_values("Machine").reset_index(drop=True)
df_plan
```
