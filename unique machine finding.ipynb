```python
import pandas as pd

file_path = "D:/Tushar/TOOL FIX/Tool Map.csv"

df = pd.read_csv(file_path)

# Convert to string & drop blanks
machines = df["Machine"].dropna().astype(str)

# Split by comma and flatten
all_machines = machines.str.split(",").explode()

# Clean spaces
all_machines = all_machines.str.strip()

# Get unique machines
unique_machines = all_machines.unique()

print("Total Unique Machines:", len(unique_machines))
print("\nMachine List:")
print(sorted(unique_machines))
```

```python
import pandas as pd

file_path = "D:/Tushar/Tool list of Mould fixation-VT.xlsx"

df = pd.read_excel(file_path)

# Convert to string & drop blanks
machines = df["Machine No."].dropna().astype(str)

# Split by comma and flatten
all_machines = machines.str.split(",").explode()

# Clean spaces
all_machines = all_machines.str.strip()

# Get unique machines
unique_machines = sorted(all_machines.unique())

print("Total Unique Machines:", len(unique_machines))
print("\nMachine List:")
print(unique_machines)

# ✅ Save to Excel
output_path = "D:/Tushar/TOOL FIX/Unique_Machines_vt.xlsx"
pd.DataFrame(unique_machines, columns=["Unique Machines"]).to_excel(output_path, index=False)

print(f"\nUnique machines saved to: {output_path}")
```
