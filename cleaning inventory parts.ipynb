```python
import pandas as pd

# Load your inventory file
file_path = "inventory.xlsx"   # change to your file name
df = pd.read_excel(file_path)

# Remove "_N" ONLY if it appears at the end
df["Material"] = df["Material"].astype(str).str.replace(r"_N$", "", regex=True)

# Save cleaned file
df.to_excel("inventory_cleaned.xlsx", index=False)

print("✅ '_N' removed from Material column where applicable.")
```
